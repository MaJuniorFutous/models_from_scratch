```python
from typing import Union

import numpy as np
```

```python
data = np.array([
    [1.0, 2.0, 3.0],
    [1.8, 2.9, 4.0],
    [2.6, 4.3, 5.0]
])
```

```python
def loss(X: np.array, y: np.array, m: Union[np.array, list], b: float):
    n, total = len(X), 0
    for i in range(n):
        total += ((m@X[i] + b) - y[i])**2

    return (total / (2*n)).item()

loss(X=data[:,:-1], y=data[:,-1:], m=[1.111, 2.222], b=1.3)
```

```python
def deriv(X: np.array, y: np.array, m: Union[np.array, list], b):
    dm, db, n = 0, 0, len(X)
    for i in range(n):
        dm += X[i]*((X[i]@m + b) - y[i])
        db += (X[i]@m + b) - y[i]

    return dm/n, db/n

deriv(X=data[:,:-1], y=data[:,-1:], m=[1.111, 2.222], b=1.3)
```

```python
def grad_descent(X: np.array, y: np.array, epochs: int = 50, lr: float = 0.01):
    m, b = np.array([0, 0]), 0
    for i in range(epochs):
        d_m, d_b = deriv(X=data[:,:-1], y=data[:,-1:], m=m, b=b)
        m = m - lr*d_m
        b = b - lr*d_b

        print(f"Epoch {i}, loss: {loss(X=X, y=y, m=m, b=b)}")

grad_descent(X=data[:,:-1], y=data[:,-1:])
```

```python
x_vals, y_vals = [i[0] for i in data], [i[1] for i in data]

print(deriv(X=x_vals, y=y_vals, m=2, b=-0.4))
```

### Visualize Loss Landscape

```python
ms = np.linspace(-5, 10, 50)
bs = np.linspace(-5, 10, 50)
losses = np.zeros((len(bs), len(ms)))

for i, b in enumerate(bs):
    for j, m in enumerate(ms):
        losses[i, j] = loss(x_vals, y_vals, m, b)
```

```python
losses.shape
```

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=[
        go.Surface(
            x=ms,
            y=bs,
            z=losses
        )
    ]
)

fig.update_layout(
    scene=dict(
        xaxis_title="Slope (m)",
        yaxis_title="Intercept (b)",
        zaxis_title="Loss"
    )
)

fig.show()
```
